```python
from pydantic import BaseModel
```

```python
class Trainers(BaseModel):  # schema

    name: str
    level: int


# create a model of above pydantic class/schema

trainer_info = {'name': 'Brock', 'level': '9'}

trainer1 = Trainers(**trainer_info) # creating an object
```

```python
# will use the above object in the function

def trainer_insert(trainer: Trainers):

    print(trainer.name)
    print(trainer.level)
    print('inserted')
```

```python
trainer_insert(trainer1)
```

```python
from typing import List, Dict, Optional
```

```python
class Trainers(BaseModel):

    name: str
    level: int = 10  # default value
    #dex: List[str]  # list of strings (if only list then use -> list)
    dex: Optional[List[str]] = None


# instantiate the pydantic object with raw dict

trainer_info = {"name": "grey", "level": "40",
    "dex": [
      "Warner",
      "Watson",
      "Gill"
    ]}

trainer1 = Trainers(**trainer_info)

def trainer_insert(trainer: Trainers):

    print(trainer.name)
    print(trainer.level)
    print(trainer1.dex)
    print('inserted')

trainer_insert(trainer1)
```
